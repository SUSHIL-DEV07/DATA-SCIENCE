Question 7 (10 Marks)

Create:

arr = np.array([11,22,33,44,55,66,77])

Using np.where():

Replace every value greater than 40 with 0.
Replace every value less than or equal to 40 with 1.

```python
A = np.array ([11,22,33,44,55,66,77])
A = np.where (A >40 , 0 , 1)
print (A)
```
